```python
import pandas as pd
df = pd.read_csv('digikala_text.csv')
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score
from sklearn.svm import SVC
model = SVC(kernel='linear')
tarjome = TfidfVectorizer()
x_train,x_test,y_train,y_test = train_test_split(df['Text'],df['Suggestion'],test_size=0.2)
x_train_int = tarjome.fit_transform(x_train)
x_test_int = tarjome.transform(x_test)
model.fit(x_train_int,y_train)
y_pred = model.predict(x_test_int)
print(accuracy_score(y_pred,y_test))

```

```python
text = input('write your exprience')
text_int = tarjome.transform([text])

y_sample = model.predict(text_int)

if y_sample[0]==1:
    print('توصیه میشه')
else :
    print('توصیه نمیشه')


```
