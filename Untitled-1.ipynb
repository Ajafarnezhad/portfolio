```python
def roar():
    print("Get out of my swamp!")
    print("I am an Ogre! Roaaar!")

# حالا بیایید تابع را صدا بزنیم (Call the function)
roar()
```

```python
def greeting(name):
    if name == "Shrek":
        print("Good morning, my love!")
    elif name == "Donkey":
        print("Donkey! Stop singing!")
    else:
        print(f"Hello, {name}. Welcome to Far Far Away.")

# تست کردن تابع با ورودی‌های مختلف
greeting("Shrek")
greeting("Donkey")
greeting("Lord Farquaad")
```

```python
def make_potion(ingredient1, ingredient2):
    # ترکیب مواد اولیه
    mixture = ingredient1 + " mixed with " + ingredient2
    return mixture

# حالا بیایید معجون بسازیم
my_potion = make_potion("Spider Eye", "Swamp Water")

# الان معجون توی متغیر my_potion ذخیره شده
print(f"Look what I have: {my_potion}")
```

```python
def calculate_smell(onion_layers):
    smell_level = onion_layers * 10
    return smell_level

current_smell = calculate_smell(5)
print(f"This onion smells like level {current_smell}")
```

```python
def eat_dinner(food="Worms"):
    print(f"Shrek is eating delicious {food}!")

# حالت پیش‌فرض (چیزی نمی‌گوییم)
eat_dinner() 

# حالت خاص (غذای جدید معرفی می‌کنیم)
eat_dinner("Eye Jelly Toast")
```

```python
def rescue_princess(hero_name, obstacle):
    if obstacle == "Dragon":
        print(f"Run {hero_name}, run!")
    else:
        print(f"{hero_name} saved the princess!")

rescue_princess("Shrek", "Dragon")
rescue_princess("Prince Charming", "Small Wall")
```
