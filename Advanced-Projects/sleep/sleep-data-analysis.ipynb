```python
# ۱. تعریف لیستی از نام ستون‌هایی که ماهیت عددی دارند
num_cols = [
    "body_weight",
    "brain_weight",
    "max_life_span",
    "gestation_time",
    "predation_index",
    "sleep_exposure_index",
    "danger_index",
    "total_sleep"
]

# ۲. تبدیل تیپ داده‌ای ستون‌ها به عدد با استفاده از یک حلقه
for col in num_cols:
    # متد pd.to_numeric برای اطمینان از عددی بودن ستون استفاده می‌شود
    # پارامتر errors="coerce": اگر مقداری غیرعددی (مثل متن یا کاراکتر خاص) مشاهده شد، 
    # آن را به NaN (مقدار خالی) تبدیل می‌کند تا اجرای کد متوقف نشود.
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

را errors="coerce"؟ در دنیای واقعی داده‌ها کثیف هستند. گاهی در یک ستون عددی، علامت سوال ? یا فاصله وجود دارد. این پارامتر از شکستن کد جلوگیری کرده و داده‌های فاسد را برای مرحله پاک‌سازی (Cleaning) آماده می‌کند.

ساختار لیست: جدا کردن نام ستون‌ها در یک متغیر به نام num_cols باعث می‌شود اگر بعداً ستون جدیدی به داده‌ها اضافه شد، فقط یک جا را تغییر دهیم.

```python
# ۳. مدیریت داده‌های گمشده (Handling Missing Values) با استراتژی میانه (Median)
# استفاده از میانه نسبت به میانگین (Mean) در داده‌های واقعی برتری دارد، 
# چون نسبت به داده‌های پرت (Outliers) حساسیت کمتری دارد.
df[num_cols] = df[num_cols].fillna(df[num_cols].median())

# ۴. بازبینی نهایی ستون‌های عددی
# نمایش چند سطر اول یا کل ستون‌های پردازش شده برای اطمینان از صحت تغییرات
df[num_cols]
```

چرا Median (میانه)؟ اگر در داده‌ها یک عدد بسیار بزرگ یا بسیار کوچک (Outlier) داشته باشیم، میانگین به شدت تغییر می‌کند، اما میانه ثابت می‌ماند و نماینده بهتری برای "مرکز داده" است.

```python
# --- مرحله مهندسی ویژگی (Feature Engineering) ---
# در این بخش از ترکیب ستون‌های موجود، متغیرهای معنادار جدیدی برای تحلیل عمیق‌تر می‌سازیم.

# ۱. شاخص‌های مغز و بدن (Brain-Body Metrics)
# اضافه کردن 1e-6 (یک عدد بسیار کوچک) برای جلوگیری از خطای تقسیم بر صفر (Zero Division Error) است.
df["brain_to_body_ratio"] = df["brain_weight"] / (df["body_weight"] + 1e-6)

# استفاده از مقیاس لگاریتمی (log1p) برای نرمال‌سازی داده‌هایی که توزیع نامتقارن (Skewed) دارند.
# log1p معادل log(1+x) است که برای مقادیر صفر هم به درستی عمل می‌کند.
df["log_body_weight"] = np.log1p(df["body_weight"])
df["log_brain_weight"] = np.log1p(df["brain_weight"])

# ۱. محاسبه نرخ خواب به ازای وزن بدن (Sleep Efficiency per Weight)
# این شاخص نشان می‌دهد که آیا حیوانات سنگین‌تر لزوماً به خواب بیشتری نیاز دارند یا خیر.
# استفاده از 1e-6 (Epsilon) برای جلوگیری از خطای DivisionByZero در صورت وجود داده صفر.
df["sleep_per_kg"] = df["total_sleep"] / (df["body_weight"] + 1e-6)

# ۲. شاخص شدت/قرارگیری در معرض خواب (Sleep Exposure Intensity)
# ترکیب میزان خواب با میزان "در معرض بودن" (Exposure)؛ 
# این متغیر بررسی می‌کند که حیوان چقدر در شرایط ناامن (بدون پناهگاه) می‌خوابد.
df["sleep_intensity_index"] = df["total_sleep"] * df["sleep_exposure_index"]

# ۳. نسبت خواب به سطح خطر (Sleep to Danger Trade-off)
# بررسی تعادل بین نیاز بیولوژیک به خواب و ریسک محیطی.
# عدد 1 در مخرج برای نرمال‌سازی (Smoothing) است تا اگر خطر صفر بود، مقدار بی‌نهایت نشود.
df["sleep_vs_danger_ratio"] = df["total_sleep"] / (df["danger_index"] + 1)

# --- مدل‌سازی شاخص‌های ریسک محیطی (Environmental Risk Modeling) ---

# ۱. محاسبه میانگین ساده سطح تهدید (Simple Threat Index)
# در اینجا فرض می‌کنیم اثر "شکارچی بودن" و "خطر کلی محیط" ارزش یکسانی دارند.
# این شاخص یک دید کلی از فشار محیطی روی گونه جانوری به ما می‌دهد.
df["threat_index"] = (df["predation_index"] + df["danger_index"]) / 2

# ۲. محاسبه نمره ریسک محیطی با استفاده از ضرایب وزنی (Weighted Risk Score)
# در دنیای واقعی، تأثیر تمام فاکتورها یکسان نیست. 
# در اینجا با اختصاص ضرایب (Weights)، اهمیت هر پارامتر را مشخص کرده‌ایم:
# - ۵۰٪ اهمیت برای شکار شدن (Predation)
# - ۳۰٪ اهمیت برای میزان در معرض خطر بودن هنگام خواب (Sleep Exposure)
# - ۲۰٪ اهمیت برای خطر کلی (Danger)
df["environment_risk_score"] = (
    0.5 * df["predation_index"] + 
    0.3 * df["sleep_exposure_index"] + 
    0.2 * df["danger_index"]
)
```

**رمال‌سازی لگاریتمی:  تفاوت وزن بین یک موش و یک فیل هزاران برابر است. با log گرفتن، این تفاوت‌ها را به مقیاسی می‌آوریم که مدل‌های یادگیری ماشین بهتر درک کنند.

پایداری عددی (Numerical Stability): تأکید کنید که استفاده از 1e-6 یک ترفند مهندسی برای جلوگیری از کرش کردن کد در صورت وجود داده‌ی صفر است.

وزن‌دهی (Weights): در متغیر environment_risk_score توضیح دهید که چطور می‌توانیم با ضرایب (0.5, 0.3, 0.2) به مدل بگوییم کدام فاکتور در محیط خطرناک‌تر است.**

شکارچی بودن (Predation) خطرناک‌تر از بقیه عوامل است، پس ضریب بیشتری به آن دادیم.

```python
# --- تحلیل استراتژی‌های زیستی و طول عمر (Life History Analysis) ---

# ۱. محاسبه بازدهی طول عمر نسبت به جرم بدن (Lifespan Efficiency per Weight)
# این شاخص بررسی می‌کند که به ازای هر واحد وزن، حیوان چقدر عمر می‌کند.
# معمولاً حیوانات کوچک‌تر نسبت به وزنشان عمر طولانی‌تری دارند (مثل خفاش در مقایسه با برخی پستانداران بزرگ).
df["lifespan_per_kg"] = df["max_life_span"] / (df["body_weight"] + 1e-6)
#جلوگیری از خطای تقسیم بر صفر (Zero Division Error) : با اضافه کردن یک عدد بسیار کوچک (1e-6) به مخرج،
# اطمینان حاصل می‌شود که حتی اگر وزن بدن صفر باشد، برنامه به جای خطا دادن، مقدار بسیار بزرگی را برمی‌گرداند که نشان‌دهنده عمر بسیار
# ۲. محاسبه نسبت دوره بارداری به کل طول عمر (Gestation-to-Lifespan Investment)
# این شاخص نشان‌دهنده میزان سرمایه‌گذاری زمانی حیوان برای تولیدمثل نسبت به کل عمرش است.
# نسبت بالا می‌تواند نشان‌دهنده استراتژی تولیدمثلی کند (K-selection) باشد، 
# جایی که حیوان زمان زیادی را صرف پرورش تعداد کمی فرزند می‌کند.
df["gestation_to_lifespan_ratio"] = df["gestation_time"] / (df["max_life_span"] + 1e-6)
```

موجودات زنده انرژی محدودی دارند. این فرمول‌ها کمک می‌کنند بفهمیم آیا یک گونه انرژی خود را صرف "رشد سریع و عمر کوتاه" می‌کند یا "رشد کند و عمر طولانی".

عادی‌سازی (Normalization): تأکید کنید که تقسیم بر وزن (body_weight) اجازه می‌دهد فیل‌ها و موش‌ها را در یک مقیاس منصفانه با هم مقایسه کنیم.

تحلیل همبستگی: این دو ویژگی جدید معمولاً همبستگی بالایی با میزان خواب دارند. حیواناتی که دوره بارداری طولانی و عمر بلندی دارند، معمولاً الگوهای خواب متفاوتی نسبت به حیوانات با چرخه زندگی سریع نشان می‌دهند.

```python
# --- محاسبه شاخص اکولوژی شناختی (Cognitive Ecology Index) ---
# این شاخص ترکیبی نشان‌دهنده تعادل میان توانمندی ذهنی، طول عمر و امنیت محیطی است.
# مقادیر به گونه‌ای تنظیم شده‌اند که بین ۰ تا ۱ باشند (Normalization).

df["cognitive_ecology_index"] = (
    # ۱. وزن ۴۰٪ برای نسبت مغز به بدن: نشان‌دهنده پتانسیل پردازش اطلاعات و هوشمندی.
    0.4 * df["brain_to_body_ratio"] + 
    
    # ۲. وزن ۳۰٪ برای طول عمر نسبی: 
    # مقادیر گمشده با میانه پر شده و سپس بر حداکثر تقسیم شده تا در بازه [0, 1] قرار گیرد.
    0.3 * (df["max_life_span"].fillna(df["max_life_span"].median()) / df["max_life_span"].max()) +
    
    # ۳. وزن ۳۰٪ برای امنیت محیطی (معکوس شاخص تهدید):
    # از '1' کسر شده چون هر چه تهدید کمتر باشد، امنیت برای فعالیت‌های شناختی بیشتر است.
    0.3 * (1 - df["threat_index"] / df["threat_index"].max())
)
```

نرمال‌سازی (Normalization): به دانشجوها توضیح دهید که نمی‌توانیم عدد "وزن مغز" را مستقیماً با "طول عمر" جمع کنیم (مثل جمع کردن سیب و پرتقال!). با تقسیم کردن هر کدام بر مقدار حداکثر (max)، همه‌چیز را به یک مقیاس مشترک بین ۰ تا ۱ تبدیل کردیم.

منطق معکوس (Inverse Logic): در بخش سوم، چرا از 1 - ... استفاده کردیم؟ چون threat_index یک پارامتر منفی (خطر) است، اما ما می‌خواهیم شاخص ما نشان‌دهنده یک امتیاز مثبت باشد. پس آن را معکوس کردیم تا "امنیت" محاسبه شود.

تفسیر خروجی: گونه‌ای که نمره نزدیکی به ۱ بگیرد، یعنی هم مغز بزرگی دارد، هم عمر طولانی و هم در محیط امنی زندگی می‌کند (مثل انسان یا برخی نخستین‌سانان). گونه‌ای که نمره نزدیکی به ۰ بگیرد، یعنی یا مغز کوچکی دارد، یا عمر کوتاهی دارد، یا در محیط خطرناکی زندگی می‌کند (مثل موش‌ها یا برخی حشرات).

```python
# --- دسته‌بندی داده‌ها و ایجاد متغیرهای کیفی (Categorical Binning) ---

# ۱. دسته‌بندی وزن بدن بر اساس چارک‌ها (Quantile-based Discretization)
# متد qcut داده‌ها را به ۴ قسمت مساوی (از نظر تعداد) تقسیم می‌کند.
# این کار کمک می‌کند تا حیوانات را بدون سوگیری، در ۴ گروه وزنی قرار دهیم.
df["size_class"] = pd.qcut(df["body_weight"], q=4, labels=["Micro", "Small", "Medium", "Large"])

# ۲. دسته‌بندی بر اساس بازه‌های مشخص برای میزان خواب (Fixed-width Binning)
# متد cut زمانی استفاده می‌شود که ما بازه‌های مشخصی (Bins) در ذهن داریم.
# در اینجا خواب را به ۴ کلاس: کوتاه، متوسط، طولانی و بسیار طولانی تقسیم کرده‌ایم.
df["sleep_profile"] = pd.cut(
    df["total_sleep"], 
    bins=[0, 6, 10, 14, 24], 
    labels=["Short Sleeper", "Moderate Sleeper", "Long Sleeper", "Extreme Sleeper"]
)

# ۳. ایجاد پروفایل ریسک بر اساس شاخص تهدید
# این دسته‌بندی به ما کمک می‌کند تا به جای کار با اعداد اعشاری، 
# حیوانات را در سه سطح خطر "کم"، "متوسط" و "زیاد" تحلیل کنیم.
df["risk_profile"] = pd.cut(
    df["threat_index"], 
    bins=[0, 2, 3.5, 5], 
    labels=["Low Risk", "Medium Risk", "High Risk"]
)
```

qcut بر اساس تعداد نمونه‌ها عمل می‌کند. یعنی در هر گروه (مثلاً Micro یا Large) تعداد تقریباً برابری از حیوانات قرار می‌گیرند. این روش برای داده‌هایی که توزیع نامتقارن دارند (مثل وزن که چند حیوان خیلی سنگین داریم) عالی است. اگر از cut استفاده کنیم، ممکن است گروه‌ها بسیار نامتوازن شوند (مثلاً یک گروه فقط شامل فیل‌ها باشد). qcut این مشکل را حل می‌کند و تضمین می‌کند که هر گروه نماینده‌ای از کل داده‌ها باشد.

در اینجا ما خودمان مرزها را تعیین می‌کنیم. مثلاً تصمیم گرفتیم کسی که زیر ۶ ساعت می‌خوابد "کم‌خواب" باشد. در این حالت ممکن است در یک گروه ۱۰۰ حیوان و در گروه دیگر فقط ۵ حیوان قرار بگیرند.

```python
# --- ایجاد ویژگی‌های تعاملی (Interaction & Trade-off Features) ---
# در علم داده، گاهی ترکیب ضربی یا تقسیمی دو متغیر، الگوهای پنهانی را آشکار می‌کند 
# که به تنهایی در داده‌ها قابل مشاهده نیستند.

# ۱. تعامل وزن مغز و میزان خواب (Brain-Sleep Interaction)
# این شاخص بررسی می‌کند که آیا حیوانات با مغز بزرگتر، برای بازیابی انرژی به خواب بیشتری نیاز دارند؟
df["brain_x_sleep"] = df["brain_weight"] * df["total_sleep"]

# ۲. تعامل اندازه بدن و خطر محیطی (Size-Danger Interaction)
# بررسی فشار محیطی بر اساس جثه؛ حیوانات بزرگتر معمولاً کمتر در معرض خطر هستند، 
# این متغیر شدتِ این رابطه را اندازه‌گیری می‌کند.
df["size_x_danger"] = df["body_weight"] * df["danger_index"]

# ۳. تعامل طول عمر و خواب (Longevity-Sleep Interaction)
# بررسی رابطه بین پتانسیل زندگی طولانی و نیاز به استراحت روزانه.
df["longevity_x_sleep"] = df["max_life_span"] * df["total_sleep"]

# ۴. شاخص توازن هوش و خطر (Intelligence-Risk Trade-off)
# این شاخص نشان می‌دهد که یک گونه به ازای هر واحد خطری که تحمل می‌کند، 
# چقدر از نظر شناختی (نسبت به جثه) توسعه یافته است.
# اضافه کردن 1e-6 برای جلوگیری از خطا در صورت صفر بودن شاخص تهدید.
df["intelligence_risk_tradeoff"] = df["brain_to_body_ratio"] / (df["threat_index"] + 1e-6)
```

اثر متقابل (Interaction Effect):ضرب دو ستون در هم، یعنی ما معتقدیم اثر متغیر اول به مقدار متغیر دوم بستگی دارد. مثلاً شاید "خواب زیاد" فقط برای "حیوانات با مغز بزرگ" معنای بیولوژیکی خاصی داشته باشد.

تحلیل توازن (Trade-off Analysis): در شاخص آخر، ما یک ویژگی مثبت (هوش) را بر یک ویژگی منفی (خطر) تقسیم کردیم. این یک روش عالی برای رتبه‌بندی گونه‌هایی است که در شرایط سخت، تکامل مغزی بالایی داشته‌اند.

log_body_weight:وزن بدن در طبیعت توزیع بسیار گسترده‌ای دارد. بدون لگاریتم، فیل‌ها در انتهای نمودار و بقیه حیوانات در یک فضای بسیار کوچک در ابتدا جمع می‌شدند. لگاریتم این فاصله را به شکلی تغییر می‌دهد که الگوها قابل دیدن باشند.

تحلیل اندازه نقاط (size): در اینجا اندازه حباب‌ها بر اساس وزن مطلق مغز است. این به ما کمک می‌کند ببیند آیا حیواناتی با مغز سنگین‌تر، لزوماً خواب بیشتر یا کمتری دارند.

تحلیل اندازه نقاط (size): در اینجا اندازه حباب‌ها بر اساس وزن مطلق مغز است. این به دانشجو کمک می‌کند ببیند آیا حیواناتی با مغز سنگین‌تر، لزوماً خواب بیشتر یا کمتری دارند.

```python
# --- بصری‌سازی رابطه وزن (لگاریتمی) و میزان خواب (Log-Scale Analysis) ---
# در این تحلیل، هدف بررسی این فرضیه است که آیا حیوانات بزرگتر کمتر می‌خوابند؟

fig = px.scatter(
    df,
    x="log_body_weight",    # استفاده از لگاریتم وزن برای نرمال‌سازی توزیع داده‌ها
    y="total_sleep",        # میزان خواب روزانه به ساعت
    color="risk_profile",   # دسته‌بندی رنگی بر اساس سطح ریسک اکولوژیک
    size="brain_weight",    # اندازه هر نقطه نشان‌دهنده وزن فیزیکی مغز است
    
    # داده‌های تکمیلی هنگام نگه داشتن موس روی نقاط
    hover_data=[
        "body_weight", "brain_weight", "brain_to_body_ratio", 
        "threat_index", "environment_risk_score", "sleep_profile"
    ],
    
    # تنظیم برچسب‌های انگلیسی برای جلوگیری از به‌هم‌ریختگی فونت
    labels={
        "log_body_weight": "Log Body Weight",
        "total_sleep": "Total Sleep (hours)",
        "risk_profile": "Risk Profile"
    },
    
    # عنوانی که هدف تحلیل را مشخص می‌کند
    title="Sleep Duration vs Body Size with Ecological Risk"
)

# نمایش نمودار تعاملی
fig.show()
```

ور خطوطی که در یک ناحیه از محور اول هستند، در محورهای بعدی هم به سمت نواحی خاصی می‌روند. مثلاً آیا حیواناتی با total_sleep بالا، لزوماً threat_index پایینی دارند؟

فیلتر کردن تعاملی: یکی از ویژگی‌های جذاب این نمودار در Plotly این است که دانشجو می‌تواند روی هر محور یک بازه را با موس انتخاب (Select) کند تا فقط خطوط مربوط به آن گروه (مثلاً حیوانات پرخطر) هایلایت شوند.

تفسیر رنگ: چون رنگ بر اساس threat_index است، خطوط بنفش و زرد به خوبی نشان می‌دهند که موجودات در معرض خطر، چه رفتاری در سایر ابعاد (مثل هوش یا میزان خواب) از خود نشان می‌دهند.

```python
# --- ترسیم نمودار مختصات موازی (Parallel Coordinates Plot) ---
# این نمودار برای تحلیل همزمان چندین متغیر عددی و کشف الگوهای پیچیده عالی است.

# ایجاد یک کپی از داده‌ها برای اطمینان از عدم تغییر در دیتاست اصلی هنگام رسم
sample = df.copy()

fig = px.parallel_coordinates(
    sample,
    # ستون‌هایی که می‌خواهیم به عنوان محورهای عمودی در نمودار نمایش داده شوند
    dimensions=[
        "log_body_weight",
        "brain_to_body_ratio",
        "total_sleep",
        "threat_index",
        "environment_risk_score",
        "cognitive_ecology_index"
    ],
    # طیف رنگی نمودار بر اساس شاخص تهدید تنظیم می‌شود
    color=sample["threat_index"],
    # استفاده از مقیاس رنگی Viridis که برای تفکیک داده‌ها بسیار استاندارد است
    color_continuous_scale="Viridis",
    # تنظیم برچسب‌های خوانا برای هر محور
    labels={
        "log_body_weight": "Log Body Weight",
        "brain_to_body_ratio": "Brain to Body Ratio",
        "total_sleep": "Total Sleep",
        "threat_index": "Threat Index",
        "environment_risk_score": "Env Risk Score",
        "cognitive_ecology_index": "Cognitive Index"
    },
    title="Multi-Dimensional Ecological and Cognitive Profiles"
)

# نمایش نمودار (در این نمودار می‌توانید با کشیدن موس روی هر محور، فیلترهای خاص اعمال کنید)
fig.show()
```

```python
# --- تحلیل فضای تهدید و مواجهه در هنگام خواب (Risk vs. Exposure Quadrant Analysis) ---
# این نمودار نشان می‌دهد که موجودات چقدر در معرض خطر هستند و محیط آن‌ها چقدر تهدیدآمیز است.

fig = px.scatter(
    df,
    x="sleep_exposure_index", # میزان بی‌دفاع بودن در زمان خواب
    y="threat_index",         # شاخص کلی تهدید در محیط
    color="risk_profile",     # دسته‌بندی رنگی بر اساس سطح ریسک
    size="body_weight",       # اندازه حباب بر اساس وزن بدن
    hover_data=[
        "total_sleep", "brain_to_body_ratio", 
        "environment_risk_score", "size_class"
    ],
    labels={
        "sleep_exposure_index": "Sleep Exposure Index",
        "threat_index": "Threat Index"
    },
    title="Sleep Exposure vs Threat Landscape"
)

# اضافه کردن خط عمودی میانه برای شاخص مواجهه در خواب 
fig.add_vline(x=df["sleep_exposure_index"].median(), line_dash="dash", line_color="gray")

# اضافه کردن خط افقی میانه برای شاخص تهدید
# تلاقی این دو خط، نمودار را به ۴ ناحیه (امن، شجاع، در معرض خطر و محافظت‌شده) تقسیم می‌کند.
fig.add_hline(y=df["threat_index"].median(), line_dash="dash", line_color="gray")

# نمایش نهایی نمودار
fig.show()
```

تحلیل چهار ناحیه (Quadrant Analysis) 

ناحیه پایین-چپ (Low Exposure / Low Threat): حیوانات امن که هم جای خواب خوبی دارند و هم دشمنان کمی دارند.

ناحیه بالا-چپ (Low Exposure / High Threat): حیواناتی که در محیط پرخطر زندگی می‌کنند اما با پیدا کردن پناهگاه‌های عالی (مثل غار)، مواجهه خود را کم کرده‌اند.

ناحیه پایین-راست (High Exposure / Low Threat): حیواناتی که جای خواب امنی ندارند اما چون دشمن جدی ندارند، بقا پیدا کرده‌اند.

ناحیه بالا-راست (High Exposure / High Threat): خطرناک‌ترین وضعیت؛ موجوداتی که هم در محیط پرخطر هستند و هم جای خواب امنی ندارند (بیشترین فشار تکاملی).

```python
# --- تحلیل توزیع خواب در کلاس‌های وزنی مختلف (Violin Plot Analysis) ---
# نمودار ویولنی ترکیبی از نمودار جعبه‌ای (Box Plot) و چگالی (Density Plot) است 
# که به ما نشان می‌دهد داده‌ها در کجا تمرکز بیشتری دارند.

fig = px.violin(
    df,
    x="size_class",    # دسته‌بندی بر اساس اندازه بدن (Micro تا Large)
    y="total_sleep",    # متغیر مورد بررسی: میزان کل خواب
    
    # اضافه کردن جعبه (Box Plot) به داخل ویولن برای نمایش چارک‌ها و میانه
    box=True, 
    
    # نمایش تمام نقاط داده (حیوانات) در کنار ویولن برای دیدن جزئیات و تراکم واقعی
    points="all", 
    
    # رنگ‌بندی مجزا برای هر کلاس وزنی جهت تفکیک بصری بهتر
    color="size_class", 
    
    # تنظیم برچسب‌های محورها به انگلیسی برای پایداری در نمایش
    labels={
        "size_class": "Size Class",
        "total_sleep": "Total Sleep (hours)"
    },
    
    # عنوان نمودار: بررسی نحوه توزیع خواب در گروه‌های مختلف جثه
    title="Sleep Distribution by Body Size Class"
)

# نمایش نمودار تعاملی
fig.show()
```

توزیع و شکل داده‌ها (Distribution Shape):  بخش‌های "شکم‌دار" یا پهن‌تر در هر ویولن، نشان‌دهنده این است که اکثر حیوانات آن گروه در آن بازه زمانی می‌خوابند.

پارامتر box=True: توضیح دهید که با این کار، ما همزمان هم آمار توصیفی (میانه و چارک‌ها) و هم شکل توزیع را یکجا می‌بینیم. این یعنی ترکیب قدرتِ دو نمودار در یک قاب.

پارامتر points="all": این بخش بسیار مهم است؛ چون به بیننده اجازه می‌دهد ببیند آیا "داده‌های پرت" (Outliers) وجود دارند یا خیر. هر نقطه نماینده یک حیوان واقعی در دیتاست است.

تحلیل خروجی: معمولاً در این نمودار مشخص می‌شود که آیا حیوانات کوچک (Micro) تنوع خواب بیشتری دارند یا حیوانات بزرگ (Large). این می‌تواند به ما بگوید که آیا اندازه بدن تأثیری بر الگوهای خواب دارد یا خیر.

```python
# --- ترسیم نمودار خورشیدی (Sunburst Chart) برای نمایش سلسله‌مراتب داده‌ها ---
# این نمودار توزیع فراوانی را در سه لایه (اندازه -> ریسک -> الگوی خواب) نمایش می‌دهد.

fig = px.sunburst(
    df,
    # تعیین مسیر سلسله‌مراتب از لایه داخلی به لایه بیرونی
    path=["size_class", "risk_profile", "sleep_profile"], 
    
    # وقتی values روی None باشد، اندازه هر قطاع بر اساس تعداد (count) رکوردها محاسبه می‌شود
    values=None, 
    
    # عنوان انگلیسی برای جلوگیری از به هم ریختگی فونت
    title="Hierarchical View of Size, Risk, and Sleep Profiles"
)

# نمایش نمودار تعاملی
# (با کلیک روی هر بخش، نمودار روی آن لایه متمرکز یا اصطلاحاً Drill-down می‌شود)
fig.show()
```

(نحوه تحلیل Sunburst):
مفهوم سهم (Proportion): به دانشجوها توضیح دهید که مساحت هر قطاع نشان‌دهنده فراوانی آن دسته است. مثلاً اگر قطاع Large بزرگتر است، یعنی تعداد حیوانات بزرگ در دیتاست بیشتر است.

آنالیز مسیر (Path Analysis): با این نمودار می‌توان به راحتی فهمید که مثلاً در دسته حیوانات "Micro"، اکثر آن‌ها در چه سطح ریسکی قرار دارند و آیا لزوماً خواب طولانی دارند یا خیر.

قابلیت Drill-down: به دانشجویان نشان دهید که اگر روی بخش Small کلیک کنند، کل نمودار دوباره حول محور آن بازسازی می‌شود تا جزئیات لایه‌های بعدی (ریسک و پروفایل خواب) را با دقت بیشتری ببینند.

کشف الگوهای غالب: این نمودار سریعاً به ما می‌گوید کدام ترکیب از ویژگی‌ها (مثلاً: Large + Low Risk + Short Sleeper) در طبیعت رایج‌تر است.

```python
# --- تحلیل همبستگی متغیرهای مهندسی شده (Correlation Heatmap) ---
# این بخش برای درک روابط خطی بین متغیرهای زیستی و اکولوژیک استفاده می‌شود.

# ۱. انتخاب لیستی از ویژگی‌های مهم برای تحلیل همبستگی
features = [
    "log_body_weight",
    "brain_to_body_ratio",
    "total_sleep",
    "threat_index",
    "environment_risk_score",
    "cognitive_ecology_index",
    "max_life_span"
]

# ۲. محاسبه ماتریس همبستگی (Correlation Matrix)
# این ماتریس اعدادی بین ۱- تا ۱+ را نشان می‌دهد که قدرت رابطه بین متغیرها را مشخص می‌کند.
corr = df[features].corr()

# ۳. ترسیم نقشه حرارتی (Heatmap) برای نمایش بصری همبستگی‌ها
fig = px.imshow(
    corr,
    text_auto=".2f",           # نمایش خودکار اعداد همبستگی با ۲ رقم اعشار روی نمودار
    color_continuous_scale="RdBu", # طیف رنگی قرمز-آبی (قرمز برای همبستگی مثبت، آبی برای منفی)
    aspect="auto",             # تنظیم خودکار ابعاد نمودار برای خوانایی بهتر
    labels=dict(color="Correlation"), # برچسب راهنمای رنگ‌ها
    title="Correlation Heatmap of Engineered Biological Features"
)

# ۴. نمایش نهایی نمودار
fig.show()
```

(تفسیر Heatmap):
مفهوم عدد همبستگی:  که عدد ۱+ یعنی رابطه مستقیم کامل، ۰ یعنی هیچ رابطه‌ای وجود ندارد، و ۱- یعنی رابطه معکوس کامل.

تفسیر رنگ‌ها (RdBu): * رنگ‌های متمایل به قرمز تیره نشان‌دهنده همبستگی مثبت قوی هستند (مثلاً احتمالاً بین هوش و طول عمر).

رنگ‌های متمایل به آبی تیره نشان‌دهنده همبستگی منفی قوی هستند (مثلاً بین وزن بدن و میزان خواب؛ یعنی هر چه حیوان سنگین‌تر باشد، کمتر می‌خوابد).

کشف الگوهای پنهان: از دانشجوها بخواهید به خانه‌هایی که رنگ‌های غلیظ دارند نگاه کنند. این‌ها همان "داستان‌هایی" هستند که داده‌ها در حال روایت کردنشان هستند.

چرا text_auto؟ ذکر کنید که نمایش اعداد روی نمودار، دقت تحلیل را بالا می‌برد و از حدس و گمان جلوگیری م
ی‌کند. این باعث می‌شود  که چقدر همبستگی وجود دارد و بتوانند آن را با سایر خانه‌ها مقایسه کنند.

```python
# --- تحلیل تقابل هوش-خطر در برابر طول عمر (Intelligence-Risk Tradeoff Analysis) ---
# این نمودار بررسی می‌کند که "هوشمند بودن در محیط‌های پرخطر" چه تاثیری بر طول عمر دارد.

fig = px.scatter(
    df,
    # محور افقی: شاخصی که از تقسیم هوش بر میزان تهدید به دست آمده (بازدهی هوش در برابر خطر)
    x="intelligence_risk_tradeoff", 
    # محور عمودی: حداکثر طول عمر گونه جانوری
    y="max_life_span", 
    # اندازه حباب: نشان‌دهنده میزان نیاز به خواب (بُعد فیزیولوژیک)
    size="total_sleep", 
    # رنگ حباب: دسته‌بندی بر اساس سطح ریسک اکولوژیک
    color="risk_profile", 
    
    # اطلاعات تکمیلی در هنگام حرکت موس روی نقاط برای تحلیل دقیق‌تر هر گونه
    hover_data=[
        "brain_to_body_ratio", "threat_index", "environment_risk_score"
    ],
    
    # تنظیم برچسب‌های انگلیسی محورها جهت جلوگیری از به هم ریختگی فونت
    labels={
        "intelligence_risk_tradeoff": "Intelligence–Risk Tradeoff",
        "max_life_span": "Max Life Span (years)"
    },
    
    # عنوان نمودار: رابطه توازن هوش و خطر با طول عمر
    title="Intelligence–Risk Tradeoff vs Longevity"
)

# نمایش نمودار تعاملی برای کشف الگوهای تکاملی
fig.show()
```

فسیر شاخص Tradeoff: اگر یک نقطه در سمت راست نمودار باشد، یعنی آن حیوان نسبت به خطراتی که تهدیدش می‌کند، مغز توسعه‌یافته‌تری دارد. این یک "مزیت تکاملی" است.

تحلیل خوشه (Cluster Analysis): بررسی کنید که آیا حیواناتی که در این شاخص (محور X) قوی هستند، عمر طولانی‌تری (محور Y) دارند؟ اگر پاسخ مثبت باشد، یعنی هوش بالا توانسته است اثر منفی خطرات محیطی را خنثی کند.

نقش خواب: اندازه حباب‌ها (میزان خواب) می‌تواند نشان دهد که آیا موجودات هوشمند و با عمر طولانی، خواب متمرکزتر و باکیفیت‌تری دارند یا خیر.

```python
# --- ساخت نمودار متحرک (Animated Scatter Plot) برای تحلیل پویای ریسک ---
# این نمودار به ما اجازه می‌دهد تا تغییرات رفتار خواب را در سطوح مختلف ریسک مشاهده کنیم.

fig = px.scatter(
    df,
    x="log_body_weight",    # محور افقی: وزن بدن در مقیاس لگاریتمی
    y="total_sleep",        # محور عمودی: میزان خواب روزانه
    
    # ایجاد فریم‌های انیمیشن بر اساس سطح خطر (Low به High)
    # با زدن دکمه Play، می‌توانید جابجایی نقاط را در سطوح مختلف ریسک ببینید.
    animation_frame="risk_profile", 
    
    size="brain_weight",    # اندازه حباب: وزن فیزیکی مغز
    color="sleep_profile",  # رنگ‌بندی: دسته‌بندی کیفی میزان خواب
    
    # نمایش اطلاعات کلیدی در هنگام نگه داشتن موس روی نقاط
    hover_data=["brain_to_body_ratio", "max_life_span"],
    
    # تنظیم برچسب‌های انگلیسی برای جلوگیری از به‌هم‌ریختگی فونت در نمودار
    labels={
        "log_body_weight": "Log Body Weight",
        "total_sleep": "Total Sleep (hours)"
    },
    
    # عنوانی که نشان‌دهنده ماهیت مقایسه‌ای نمودار است
    title="Sleep vs Body Size Across Risk Profiles"
)

# تنظیمات اضافی برای انیمیشن (سرعت و نحوه انتقال بین فریم‌ها)
fig.layout.updatemenus[0].buttons[0].args[1]['frame']['duration'] = 1000

# نمایش نمودار انیمیشنی
fig.show()
```

بعد زمان/گروه: ب animation_frame مانند این است که برای هر سطح ریسک یک نمودار جداگانه داشته باشیم، اما همه‌ی آن‌ها را در یک اسلاید متحرک ترکیب کرده‌ایم تا "روند" (Trend) تغییرات بهتر درک شود.

ثبات محورها: تأکید کنید که در انیمیشن، محورهای $X$ و $Y$ ثابت می‌مانند؛ این ویژگی باعث می‌شود دانشجو متوجه شود که با افزایش ریسک، نقاط به سمت پایین (خواب کمتر) یا بالا (خواب بیشتر) حرکت می‌کنند.

تحلیل بصری:  "وقتی انیمیشن از Low Risk به High Risk می‌رود، آیا نقاط با اندازه بزرگ (مغز سنگین) غیب می‌شوند یا جایشان تغییر می‌کند؟" این کار باعث درگیر شدن ذهن آن‌ها با منطق تکاملی داده‌ها می‌شود.

```python
# --- تحلیل ترکیب‌بندی اکولوژیک (Ecological Composition Analysis) ---
# در این بخش بررسی می‌کنیم که در هر گروه وزنی، چه درصدی از حیوانات در خطر هستند.

census_df = df.groupby(["size_class", "risk_profile"]).size().reset_index(name="count")

fig = px.bar(
    census_df,
    x="size_class",
    y="count",
    color="risk_profile",
    title="Population Distribution: Size Class vs. Risk Profile",
    barmode="stack", # قرار دادن گروه‌های ریسک روی هم برای مقایسه کل
    labels={"count": "Number of Species", "size_class": "Size Category"},
    color_discrete_sequence=px.colors.qualitative.Pastel
)

fig.show()
```

تشخیص خطی بودن: هرجا که نقاط به شکل یک خط یا بیضی کشیده دیده می‌شوند، نشان‌دهنده یک همبستگی قوی است.

اهمیت تمیزکاری (Pretty Labels):  در گزارش‌های مدیریتی و علمی، نام‌های فنی (مثل log_body_weight) باید به نام‌های قابل فهم برای انسان تبدیل شوند.

```python
# --- خلاصه آماری نهایی و خروجی گرفتن (Final Summary & Export) ---

# ۱. ایجاد یک جدول خلاصه از میانگین شاخص‌های مهندسی شده برای هر پروفایل ریسک
# این جدول "داستان نهایی" داده‌های شما را به صورت عددی بیان می‌کند.
summary_table = df.groupby("risk_profile")[[
    "total_sleep", 
    "brain_to_body_ratio", 
    "cognitive_ecology_index",
    "max_life_span"
]].mean().round(2)

print("--- Final Summary Table (Averages per Risk Level) ---")
print(summary_table)

# ۲. ذخیره دیتاست نهایی شامل تمام ستون‌های جدید و مهندسی شده
# این فایل برای استفاده در مدل‌های Machine Learning یا ارائه در اکسل آماده است.
df.to_csv("mammals_enriched_data.csv", index=False)

print("\nSuccess! Final dataset saved as 'mammals_enriched_data.csv'.")
```

```python
# --- شناسایی گونه‌های خاص و داده‌های پرت (Outlier Detection) ---
# در این بخش، حیواناتی که رفتار آن‌ها با بقیه تفاوت فاحشی دارد را شناسایی می‌کنیم.
# از شاخص Z-Score برای پیدا کردن مقادیری که بیش از 2 انحراف معیار با میانگین فاصله دارند استفاده می‌کنیم.

from scipy import stats

# محاسبه Z-Score برای ستون خواب
df['sleep_z_score'] = np.abs(stats.zscore(df['total_sleep']))

# جدا کردن حیواناتی که خواب آن‌ها بسیار غیرعادی است (Outliers)
outliers = df[df['sleep_z_score'] > 2]

fig = px.scatter(
    outliers,
    x="body_weight",
    y="total_sleep",
    text="body_weight", # نمایش نام یا مقدار روی نقاط
    title="Biological Outliers: Species with Unusual Sleep Patterns",
    labels={"body_weight": "Body Weight", "total_sleep": "Total Sleep"},
    color_discrete_sequence=['red']
)

fig.update_traces(textposition='top center')
fig.show()
```

حیواناتی که رفتارهای خیلی عجیب دارن (مثلاً نسبت به وزن‌شون خیلی زیاد می‌خوابن) رو شناسایی می‌کنیم.

```python
# --- تحلیل تراکم حیات (Biological Density Analysis) - نسخه اصلاح شده ---

fig = px.density_heatmap(
    df,
    x="log_body_weight",
    y="max_life_span",
    marginal_x="rug", 
    marginal_y="histogram",
    labels={
        "log_body_weight": "Log Body Weight",
        "max_life_span": "Max Life Span"
    },
    title="Density Mapping: Where do most species fall?"
)

# اعمال مقیاس رنگی به صورت مستقیم روی Trace اصلی برای جلوگیری از تداخل با حاشیه‌ها
fig.update_traces(colorscale="Viridis", selector=dict(type='histogram2d'))

fig.show()
```

```python
# --- گزارش آماری نهایی (Scientific Summary Report) ---
# در این بخش، میانگین ویژگی‌های کلیدی را برای هر کلاس وزنی محاسبه می‌کنیم.

summary_report = df.groupby("size_class", observed=True).agg({
    "total_sleep": "mean",
    "brain_to_body_ratio": "mean",
    "cognitive_ecology_index": "mean",
    "max_life_span": "mean"
}).round(3) # گرد کردن اعداد تا ۳ رقم اعشار برای خوانایی بیشتر

# نمایش جدول در خروجی (بسیار مفید برای اسلایدهای ارائه)
print("Detailed Biological Metrics by Size Class:")
display(summary_report)
```

```python
# --- نمودار راداری: مقایسه استراتژی بقا در کلاس‌های وزنی (نسخه اصلاح شده) ---

# ۱. آماده‌سازی و نرمال‌سازی داده‌ها
radar_df = df.groupby("size_class", observed=True)[["total_sleep", "threat_index", "brain_to_body_ratio"]].mean().reset_index()
for col in ["total_sleep", "threat_index", "brain_to_body_ratio"]:
    radar_df[col] = radar_df[col] / radar_df[col].max()

# ۲. رسم نمودار
fig = px.line_polar(
    radar_df, 
    r="total_sleep", 
    theta="size_class", 
    line_close=True,
    title="Biological Radar: Strategy by Size Class",
    template="plotly_dark"
)

# ۳. پر کردن داخل نمودار برای تاکید بیشتر بر تفاوت‌ها
fig.update_traces(fill='toself') 

fig.show()
```
